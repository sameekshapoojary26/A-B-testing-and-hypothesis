```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
total_passengers = len(df)
survived = df["Survived"].sum()
not_survived = total_passengers - survived
survival_rate = (survived / total_passengers) * 100
average_fare = df["Fare"].mean()

print("TOTAL PASSENGERS:", total_passengers)
print("SURVIVED:", survived)
print("NOT SURVIVED:", not_survived)
print("SURVIVAL RATE:", round(survival_rate, 2), "%")
print("AVERAGE FARE:", round(average_fare, 2))
```

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Select dataset
dataset_names = list(datasets.keys())

dataset_dropdown = widgets.Dropdown(
    options=dataset_names,
    description="Dataset:"
)

output = widgets.Output()

def create_dashboard(change=None):
    with output:
        clear_output(wait=True)

        df_dash = datasets[dataset_dropdown.value].copy()

        print("📊 INTERACTIVE BUSINESS DASHBOARD")
        print("Dataset:", dataset_dropdown.value)

        # KPI cards
        numeric = df_dash.select_dtypes(include="number")
        cols = numeric.columns.tolist()

        cards = [
            ("Total Records", f"{len(df_dash):,}"),
            ("Numeric Fields", str(len(cols))),
            ("Missing Values", f"{df_dash.isnull().sum().sum():,}"),
            ("Total Numeric Value", f"{numeric.sum().sum():,.2f}" if len(cols) else "N/A")
        ]

        html = "<div style='display:flex;gap:12px;flex-wrap:wrap'>"
        for title, value in cards:
            html += f"""
            <div style='background:#eaf2f8;padding:18px;
                        border-radius:10px;min-width:140px'>
                <b>{title}</b><h2>{value}</h2>
            </div>"""
        html += "</div>"

        display(HTML(html))

        # Numeric charts
        if len(cols) >= 1:
            chart_col = widgets.Dropdown(
                options=cols,
                description="Chart:"
            )

            def show_chart(col):
                fig = px.histogram(
                    df_dash, x=col,
                    title=f"Distribution of {col}"
                )
                fig.show()

            widgets.interact(show_chart, col=chart_col)

        # Dataset preview
        print("Dataset Preview")
        display(df_dash.head(10))

dataset_dropdown.observe(create_dashboard, names="value")

display(dataset_dropdown)
display(output)

create_dashboard()
```
